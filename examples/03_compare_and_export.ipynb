# Compare, export, and review figures

This notebook creates a small set of related figures for the comparison workflow. Run all cells, select several thumbnails in Clio, and open comparison mode.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(11)
x = np.linspace(0, 10, 180)
```

```python
# figure: Baseline model
# tags: comparison, model, baseline
baseline = 0.7 * x + 2 + rng.normal(0, 0.8, x.size)
plt.figure(figsize=(6.5, 3.6))
plt.scatter(x, baseline, s=16, alpha=0.7)
plt.plot(x, 0.7 * x + 2, color='black', linewidth=2)
plt.title('Baseline fit')
plt.xlabel('input')
plt.ylabel('prediction')
plt.show()
```

```python
# figure: Seasonal model
# tags: comparison, model, seasonal
seasonal = baseline + 1.2 * np.sin(1.5 * x)
plt.figure(figsize=(6.5, 3.6))
plt.scatter(x, seasonal, s=16, alpha=0.7, color='tab:green')
plt.plot(x, 0.7 * x + 2 + 1.2 * np.sin(1.5 * x), color='black', linewidth=2)
plt.title('Seasonal fit')
plt.xlabel('input')
plt.ylabel('prediction')
plt.show()
```

```python
# figure: Baseline residuals
# tags: comparison, residuals, baseline
plt.figure(figsize=(6.5, 3.6))
plt.scatter(x, baseline - (0.7 * x + 2), s=16, alpha=0.7, color='tab:blue')
plt.axhline(0, color='black', linewidth=1)
plt.title('Baseline residuals')
plt.xlabel('input')
plt.ylabel('residual')
plt.show()
```

```python
# figure: Seasonal residuals
# tags: comparison, residuals, seasonal
plt.figure(figsize=(6.5, 3.6))
plt.scatter(x, seasonal - (0.7 * x + 2), s=16, alpha=0.7, color='tab:purple')
plt.axhline(0, color='black', linewidth=1)
plt.title('Seasonal residuals')
plt.xlabel('input')
plt.ylabel('residual')
plt.show()
```
