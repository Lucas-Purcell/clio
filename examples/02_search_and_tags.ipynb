```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
time = np.linspace(0, 24, 120)
```

```python
# figure: Daily temperature profile
# tags: climate, temperature, line, draft
temperature = 14 + 7 * np.sin((time - 7) * np.pi / 12) + rng.normal(0, 0.7, time.size)
plt.figure(figsize=(7, 3.8))
plt.plot(time, temperature, color='tab:orange')
plt.xlabel('hour')
plt.ylabel('temperature (°C)')
plt.title('Synthetic daily temperature')
plt.grid(alpha=0.25)
plt.show()
```

```python
# figure: Temperature residuals
# tags: climate, temperature, residuals, quality-check
trend = 14 + 7 * np.sin((time - 7) * np.pi / 12)
residuals = temperature - trend
plt.figure(figsize=(7, 3.8))
plt.hist(residuals, bins=18, color='tab:blue', edgecolor='white')
plt.axvline(0, color='black', linewidth=1)
plt.xlabel('residual (°C)')
plt.title('Residual distribution')
plt.show()
```

```python
# figure: Sample path, Smoothed sample path
# tags: signal, multi-output, moving-average
signal = np.cumsum(rng.normal(size=90))
window = 9
smooth = np.convolve(signal, np.ones(window) / window, mode='same')

plt.figure(figsize=(7, 3.2))
plt.plot(signal, color='tab:gray')
plt.title('Raw random-walk signal')
plt.show()

plt.figure(figsize=(7, 3.2))
plt.plot(signal, color='lightgray', label='raw')
plt.plot(smooth, color='tab:green', linewidth=2, label='moving average')
plt.legend()
plt.title('Smoothed random-walk signal')
plt.show()
```
