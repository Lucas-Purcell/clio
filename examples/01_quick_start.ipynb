This notebook is the smallest complete Clio demonstration. It works unchanged in JupyterLab and VS Code. If needed, install the demo dependencies with `python -m pip install numpy matplotlib`. Run all code cells, then open Clio.

To install the jupyter lab version of the extension type `python -m pip install clio-jupyter numpy matplotlib
jupyter lab` in the terminal.

- **JupyterLab:** open the Clio sidebar or run **Clio: Open Gallery as Tab**.
- **VS Code:** select the Clio icon in the Activity Bar, or run **Clio: Open Gallery in Editor**.

Clio automatically finds PNG plot outputs; no manual scan is needed for open notebooks.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
x = np.linspace(0, 2 * np.pi, 240)
```

```python
# figure: Sine and cosine
# tags: quickstart, line, trigonometry
plt.figure(figsize=(7, 4))
plt.plot(x, np.sin(x), label='sin(x)', linewidth=2)
plt.plot(x, np.cos(x), label='cos(x)', linewidth=2)
plt.title('Two familiar functions')
plt.xlabel('x')
plt.ylabel('value')
plt.legend()
plt.grid(alpha=0.25)
plt.show()
```

```python
# figure: Noisy measurements
# tags: quickstart, scatter, measurements
sample_x = rng.uniform(0, 10, 100)
sample_y = 0.6 * sample_x + rng.normal(0, 1.1, sample_x.size)
plt.figure(figsize=(7, 4))
plt.scatter(sample_x, sample_y, c=sample_y, cmap='viridis', s=35)
plt.colorbar(label='measurement')
plt.title('A small measurement set')
plt.xlabel('input')
plt.ylabel('response')
plt.show()
```
